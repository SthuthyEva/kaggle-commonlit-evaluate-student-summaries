```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
exp_name = "d3b9-f4-s42-e43-b16-d0-2e5-1e3-wd005-wm0-m3-x512"
label_cols = ["cohesion", "syntax", "vocabulary", "phraseology", "grammar", "conventions"]

folds = 4

rows = 1
cols = len(label_cols)

for idx, fold in enumerate(range(folds)):
    print("Fold 1 : ", fold)
    fig, axes = plt.subplots(rows, cols, figsize=(cols*6, rows*6))
    df = pd.read_csv(f"../../output/feedback-prize-english-language-learning/{exp_name}/predictions_fold{fold}.csv")
#     display(df.head())
    for idx1, col in enumerate(label_cols):
#         print(col)
#         sns.histplot(data=df, x=col, ax=axes[0, idx1])
        sns.histplot(data=df, x=f'{col}_predictions', ax=axes[idx1])
    
```
